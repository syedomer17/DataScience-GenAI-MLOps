```python
# drop the unname column because it does not have any values 
df = df.drop("Unnamed: 0",axis = True)
```

```python
df.head()
```

```python
# change the weelstudyHouser columns 
df["WklyStudyHours"] = df["WklyStudyHours"].str.replace("05-Oct","5-10")
```

```python
df.head()
```

```python
plt.figure(figsize=(5,5))
ax = sns.countplot(data = df, x = "Gender")
ax.bar_label(ax.containers[0])
plt.title("Male vs Female")
plt.show()
```

```python
gb = df.groupby("ParentEduc").agg({
    "MathScore": "mean",
    "ReadingScore": "mean",
    "WritingScore": "mean"
})
```

```python
gb
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(gb,annot=True)
plt.title("Relation between parent and Child Score")
plt.show()
```

```python
marrySataus = df.groupby("ParentMaritalStatus").agg({
    "MathScore": "mean",
    "ReadingScore": "mean",
    "WritingScore": "mean"
})
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(marrySataus,annot=True)
plt.show()
```
